Combined L1 and L2 tests

```python
import numpy as np
from mlp.penalties import L1Penalty, L2Penalty, L1L2MixPenalty

rng = np.random.RandomState(12345)
parameter = rng.normal(loc=0., scale=3.0, size=(32, 8))
coefficient = 1e-4

mix_as_l1 = L1L2MixPenalty(coefficient, mixing_term=1.0)
mix_as_l2 = L1L2MixPenalty(coefficient, mixing_term=0.0)
pure_l1 = L1Penalty(coefficient)
pure_l2 = L2Penalty(coefficient)

mix_l1_value = mix_as_l1(parameter)
pure_l1_value = pure_l1(parameter)
mix_l1_grad = mix_as_l1.grad(parameter)
pure_l1_grad = pure_l1.grad(parameter)

mix_l2_value = mix_as_l2(parameter)
pure_l2_value = pure_l2(parameter)
mix_l2_grad = mix_as_l2.grad(parameter)
pure_l2_grad = pure_l2.grad(parameter)

assert np.allclose(mix_l1_value, pure_l1_value), (
    f'L1 equivalence failed: mix value {mix_l1_value} vs pure {pure_l1_value}'
)
assert np.allclose(mix_l2_value, pure_l2_value), (
    f'L2 equivalence failed: mix value {mix_l2_value} vs pure {pure_l2_value}'
)
assert np.allclose(mix_l1_grad, pure_l1_grad), (
    f'L1 grad mismatch max diff {np.abs(mix_l1_grad - pure_l1_grad).max()}'
)
assert np.allclose(mix_l2_grad, pure_l2_grad), (
    f'L2 grad mismatch max diff {np.abs(mix_l2_grad - pure_l2_grad).max()}'
)
print('L1L2MixPenalty behaves like L1 when mixing_term=1 (value & grad match).')
print('L1L2MixPenalty behaves like L2 when mixing_term=0 (value & grad match).')
```
